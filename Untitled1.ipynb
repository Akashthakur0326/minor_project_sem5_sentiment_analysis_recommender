```python
import pandas as pd
import torch
from transformers import AutoTokenizer, AutoModelForSequenceClassification
from google.colab import drive

drive.mount('/content/drive')

FILE_PATH = "/content/drive/Othercomputers/My laptop/Desktop/minor_poject_sem5/data/mobiles_final (2).csv"
SAVE_PATH = "/content/drive/Othercomputers/My laptop/Desktop/minor_poject_sem5/final_data.csv"

df = pd.read_csv(FILE_PATH)
print("Loaded DF:", df.shape)

# Load BERTweet sentiment model
MODEL_NAME = "finiteautomata/bertweet-base-sentiment-analysis"

tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)
model = AutoModelForSequenceClassification.from_pretrained(MODEL_NAME)
labels = ["negative", "neutral", "positive"]

device = "cuda" if torch.cuda.is_available() else "cpu"
model.to(device)
print("Model loaded on:", device)
```

```python
def classify_sentiment(text):
    if not isinstance(text, str) or text.strip() == "":
        return None

    inputs = tokenizer(
        text,
        return_tensors="pt",
        truncation=True,
        padding=True,
        max_length=128
    ).to(device)

    with torch.no_grad():
        outputs = model(**inputs)
        probs = torch.softmax(outputs.logits, dim=1)[0]
        label_id = torch.argmax(probs).item()
        return labels[label_id]
```

```python
# Collect all tweet columns (Tweet_1 ... Tweet_10 + Gen_Tweet_1...3)
tweet_cols = [c for c in df.columns if "Tweet_" in c]
print("Tweet columns:", tweet_cols)

# New score columns
df["Sentiment_Positive_Score"] = 0.0
df["Sentiment_Negative_Score"] = 0.0
df["Sentiment_Neutral_Score"]  = 0.0
df["Sentiment_Weighted_Score"] = 0.0     # final sentiment value per product

def classify_sentiment_scores(text):
    if not isinstance(text, str) or text.strip() == "":
        return (0, 0, 0)   # no text → no score

    inputs = tokenizer(
        text,
        return_tensors="pt",
        truncation=True,
        padding=True,
        max_length=128
    ).to(device)

    with torch.no_grad():
        logits = model(**inputs).logits[0]
        probs = torch.softmax(logits, dim=0).cpu().numpy()

    # probs = [neg, neu, pos]
    neg, neu, pos = probs
    return pos, neg, neu


for idx, row in df.iterrows():
    total_pos = total_neg = total_neu = 0.0
    weighted_sum = 0.0
    count = 0

    for col in tweet_cols:
        pos, neg, neu = classify_sentiment_scores(row[col])

        total_pos += pos
        total_neg += neg
        total_neu += neu

        # Weighted sentiment score:
        # +1 * positive_prob  + 0 * neutral_prob + (-1) * negative_prob
        weighted_sum += (pos * 1) + (neg * -1)

        # count only if tweet exists
        if isinstance(row[col], str) and row[col].strip() != "":
            count += 1

    df.at[idx, "Sentiment_Positive_Score"] = total_pos
    df.at[idx, "Sentiment_Negative_Score"] = total_neg
    df.at[idx, "Sentiment_Neutral_Score"]  = total_neu

    # average score (avoid divide-by-zero)
    if count > 0:
        df.at[idx, "Sentiment_Weighted_Score"] = weighted_sum / count
    else:
        df.at[idx, "Sentiment_Weighted_Score"] = 0

    if idx % 50 == 0:
        print(f"Processed row {idx}")
```

```python
df.info()
```

```python
# Number of tweet columns available
MAX_TWEETS = 10

# Create final column
df["Final_Sentiment"] = df.apply(
    lambda row: row["Sentiment_Weighted_Score"] * (row["Tweet_Count"] / MAX_TWEETS)
    if pd.notna(row["Sentiment_Weighted_Score"]) else None,
    axis=1
)

print("Final sentiment column created.")

# Save
save_path = "/content/drive/Othercomputers/My laptop/Desktop/minor_poject_sem5/final_data.csv"
df.to_csv(save_path, index=False)

print("Saved:", save_path)
```
